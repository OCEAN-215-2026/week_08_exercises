### Workflow

Each question will be timed to ensure that everyone gets to work on at least a part of every question part. When working on the exercises, you should alternate between the role of "driver" and "navigator", using generative AI to fill in the remaining role. The way you work with generative AI in each of the role is as follows:

- **Driver:** as a driver, your primary task is the detailed implementations of the coding solution. You will make a first attempt at producing the codes required to complete the question part. If you are stuck or if the code is bugged, you can consult internet resources or generative AI for help. However, you should not request complete solutions to the assigned exercise. In addition, you should make sure that you can explain the codes generated by the AI or referenced on the internet in your own words.

- **Navigator:** the navigator is responsible for navigating the bigger picture of the problem. The navigator will ask AI to make a first attempt at producing the codes required to complete the question part (this is the only instance in the entire course where direct solution generation from AI is allowed). The navigator should ask for clarification from the generative AI if any part of the generated codes are unclear to the navigator. If the generated code does not fully solve the problem at hand or is unsatisfactory for other reasons (e.g., violation of best coding practices), the navigator should make suggestions to the generative AI on how the codes may be modified and ask for updated codes. The navigator should preserve the overall structure of the AI-generated solution, but may make modifications necessary to correct errors, improve clarity, or better satisfy the requirements.


The roles of driver and navigator(s) should be rotated for **every subpart of the question**.
